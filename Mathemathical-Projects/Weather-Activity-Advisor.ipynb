```python
temp = 20
is_sunny = False

if temp >= 30 and is_sunny:
    print("It's SO SUNNY for outside walk.")
elif 20 <= temp < 30 and is_sunny:
    print("This is the great sunny time for walk")
elif 20 <= temp < 30 and not is_sunny:
    print("It's a normal and little hot day.")
elif 10 <= temp < 20 and is_sunny:
    print("This is the great sunny time for walk")
elif 10 <= temp < 20 and not is_sunny:
    print("This is the great cloudy time for walk")
else:
    print("Weather is extreme, stay inside!")
```
