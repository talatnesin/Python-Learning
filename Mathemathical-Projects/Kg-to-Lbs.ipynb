```python
weight = float(input("Enter weight:"))
unit = input("Kilograms or Paunds?: (K/P)")

if unit == "K":
    weight = weight * 2.205
    unit = "Kgs."
    print(f"Your weight is {round(weight, 3)} {unit}")

elif unit == "P":
    weight = weight / 2.205
    unit = "Lbs."
    print(f"Your weight is {round(weight, 3)} {unit}")

else:
    print(f"Please select K or P, {unit} it's not a command.")
```
