```python
print("Welcome the Celcius - Fahreneit converter!")

unit = input("Is this tempature in Celcius or Fahreneit? (C/F)")
temp = float(input("Enter the tempature:"))

if unit == "C":
    temp = (temp * 9 / 5) + 32
    print(temp)

elif unit == "F":
    temp = (temp - 32) * 5 / 9
    print(temp)

else:
    print(f"You should write C or F, not {unit}")
```
