```python
print("(You will only write to 2 area. For example: 9 + 9 is okay but 9 + 9 + 9 is NOT okay.)")
operator = input("select a operator (+, -, *, /)")
num1 = float(input("Enter the first number:"))
num2 = float(input("Enter the second number"))

if operator == "+":
   result = num1 + num2
   print(result)
elif operator == "-":
   result = num1 - num2
   print(result)
elif operator == "*":
   result = num1 * num2
   print(result)
elif operator == "/":
   result = num1 / num2
   print(round(result, 3))
else:
   print("Please select + or - or * or /")
```
